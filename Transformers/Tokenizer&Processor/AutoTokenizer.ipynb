# tokenizer(sequence)

```python
inputs = tokenizer(
    sequences[0],  # 单个句子
    truncation=True,  # 超出max_length截断处理
    padding=True,  # 填充方式选择 [True, 'longest', 'max_length', 'do_not_pad']
    max_length=8192,  # 最长长度,不设置默认为模型最大长度
    add_special_tokens=True,  # text添加特殊key
    return_length=True,  # 返回有效长度
    return_overflowing_tokens=False,  # 返回所有的文本片段（由于文本比较长，默认情况下超过预设截断长度的token会被丢失。如果设置了return_overflowing_tokens=True则会返回所有的token片段）。
    return_tensors="pt",  # 返回数据格式 np pt tf jax
)
print(inputs)
print(inputs["input_ids"])  # 对应文字id
print(inputs["attention_mask"])  # 对应是否是文字
print(inputs["length"])  # 对应总长度长度
```

```python
tokenizer.decode(inputs["input_ids"], skip_special_tokens=True)
```

# apply_chat_template

```python
help(tokenizer.apply_chat_template)
```

```python
conversation = [
    {"role": "system", "content": "You are a helpful assistant."},
    {"role": "user", "content": "What is the capital of France?"},
    {"role": "assistant", "content": "The capital of France is Paris."},
    {"role": "user", "content": "Introduce some foods in Paris."},
]
```

```python
text_with_template = tokenizer.apply_chat_template(
    conversation=conversation,
    add_generation_prompt=False,
    tokenize=False,
    padding=False,
    truncation=False,
    max_length=None,
    return_tensors="pt",
)
print(text_with_template)
```

```python
text_with_template = tokenizer.apply_chat_template(
    conversation=conversation,
    add_generation_prompt=True,  # 添加提示模型生成后续内容的特殊字符
    tokenize=False,
    padding=False,
    truncation=False,
    max_length=None,
    return_tensors="pt",
)
print(text_with_template)
```

```python
tokenizer.apply_chat_template(
    conversation=conversation,
    add_generation_prompt=True,
    tokenize=True,
    padding=False,
    truncation=False,
    max_length=None,
    return_tensors="pt",
)
```

# encode_message_with_chat_template

分词单个消息。此方法是 apply_chat_template 的便捷包装，允许您逐条分词消息。这对于 token-by-token 流式传输等场景非常有用。此方法不能保证完美。对于某些模型，可能无法可靠地分词单个消息。例如，如果聊天模板在每条消息后添加 token，但还有一个添加到整个聊天的前缀，那么就无法区分聊天开始 token 和消息开始 token。在这些情况下，此方法会尽力找到正确的分词，但可能不完美。注意：此方法不支持 add_generation_prompt。如果您想添加生成提示，应在分词聊天记录后单独进行。


```python
tokenizer.encode_message_with_chat_template(conversation[-1], conversation[:-1])
```

```python
tokenizer.apply_chat_template(
    conversation=[conversation[-1]],
    add_generation_prompt=True,
    tokenize=True,
    padding=False,
    truncation=False,
    return_tensors="pt",
)["input_ids"]
```
