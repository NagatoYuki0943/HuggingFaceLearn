# apply_chat_template

```python
conversation = [
    {"role": "system", "content": "你是一个专业的图像描述工具，你需要将图片中的内容描述为中文。"},
    {"role": "user", "content":
        [
            {"type": "text", "text": "我这里有两张图片:"},
            {
                "type": "image",
                "image": "./images/image1.jpg",
            },
            {"type": "text", "text": "上面的图片是一只绿色小精灵，"},
            {
                "type": "image",
                "image": "./images/image2.jpg",
            },
            {"type": "text", "text": "上面的图片是一只绿色小虫子，请根据上面的两张图片中的内容，写一个小故事"},
        ]
    },
    {"role": "assistant", "content": "好的，请稍等，我正在为您生成一篇小故事。"},
]
```

```python
text_with_template = tokenizer.apply_chat_template(
    conversation=conversation,
    add_generation_prompt=False,
    tokenize=False,
    padding=False,
    truncation=False,
    max_length=None,
    return_tensors="pt",
)
print(text_with_template)
```

```python
inputs = tokenizer.apply_chat_template(
    conversation=conversation,
    add_generation_prompt=True,
    tokenize=True,
    padding=False,
    truncation=False,
    max_length=None,
    return_tensors="pt",
)
inputs
```

# 查找对应的 token_id

```python
conversation = [
    {
        "role": "user",
        "content": [
            {
                "type": "image",
                "image": "./images/image1.jpg",
            },
            {"type": "text", "text": "描述一下上面图片的内容"},
        ],
    }
]
```

```python
text_with_template = tokenizer.apply_chat_template(
    conversation=conversation,
    add_generation_prompt=False,
    tokenize=False,
    padding=False,
    truncation=False,
    max_length=None,
    return_tensors="pt",
)
text_with_template
```

```python
inputs = tokenizer.apply_chat_template(
    conversation=conversation,
    add_generation_prompt=True,
    tokenize=True,
    padding=False,
    truncation=False,
    max_length=None,
    return_tensors="pt",
)
inputs
```

```python
for input_id in inputs["input_ids"][0]:
    print(input_id, "->", repr(tokenizer.decode(input_id)))
```
