# Column Linear

<img src="Tensor Parallel Column Linear.png" alt="Column Linear" width="800">

```python
x = np.array([
    [0, 1],
    [2, 3],
    [4, 5],
    [6, 7],
])
```

```python
w = np.array([
    [10, 30],
    [20, 40],
])
```

```python
w_0 = w[:, 0:1]
w_0
```

```python
w_1 = w[:, 1:2]
w_1
```

```python
y = x @ w
y
```

```python
y_0 = x @ w_0
y_0
```

```python
y_1 = x @ w_1
y_1
```

```python
y_full = np.concat([y_0, y_1], axis=1)
y_full
```

```python
y == y_full
```

# Row Linear

<img src="Tensor Parallel Row Linear.png" alt="Row Linear" width="800">

```python
x = np.array([
    [0, 1],
    [2, 3],
    [4, 5],
    [6, 7],
])
```

```python
x_0 = x[:, 0:1]
x_0
```

```python
x_1 = x[:, 1:2]
x_1
```

```python
w = np.array([
    [10, 30],
    [20, 40],
])
```

```python
w_0 = w[0:1, :]
w_0
```

```python
w_1 = w[1:2, :]
w_1
```

```python
y = x @ w
y
```

```python
y_0 = x_0 @ w_0
y_0
```

```python
y_1 = x_1 @ w_1
y_1
```

```python
y_full = y_0 + y_1
y_full
```

```python
y == y_full
```

# Column Linear + Row Linear

<img src="Tensor Parallel Column Linear + Row Linear.png" alt="Column Linear + Row Linear" width="1000">

```python
x = np.array([
    [0, 1],
    [2, 3],
    [4, 5],
    [6, 7],
])
```

```python
w1 = np.array([
    [10, 30],
    [20, 40],
])
```

```python
w1_0 = w1[:, 0:1]
w1_0
```

```python
w1_1 = w1[:, 1:2]
w1_1
```

```python
w2 = np.array([
    [10, 30],
    [20, 40],
])
```

```python
w2_0 = w2[0:1, :]
w2_0
```

```python
w2_1 = w2[1:2, :]
w2_1
```

```python
y2 = x @ w1 @ w2
y2
```

```python
y1_0 = x @ w1_0
y1_0
```

```python
y2_0 = y1_0 @ w2_0
y2_0
```

```python
y1_1 = x @ w1_1
y1_1
```

```python
y2_1 = y1_1 @ w2_1
y2_1
```

```python
y2_full = y2_0 + y2_1
y2_full
```

```python
y2 == y2_full
```
